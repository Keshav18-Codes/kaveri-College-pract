```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()


df= pd.DataFrame(iris.data,columns=iris.feature_names)


df["target"]=iris.target


print("DataFrame:")
print(df.head())


x=iris.data
y=iris.target

scaler= StandardScaler()
x_scaled = scaler.fit_transform(x)


pca = PCA(n_components=2)
x_pca = pca.fit_transform(x_scaled)

print("\n Data after PCA : ")
print(x_pca[:5])


```

```python
x_train, x_test, y_train, y_test = train_test_split(x_pca,y, test_size=0.2, random_state=42)
```

```python
model= KNeighborsClassifier(n_neighbors=3)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("\n Accuracy :",accuracy)
```

```python
new_flower = [[5.2,3.5,1.4,0.2]]
```

```python
new_scaled = scaler.transform(new_flower)
```

```python
new_pca = pca.transform(new_scaled)
```

```python
prediction = model.predict(new_pca)
```

```python
print("\n Predicted flower class :",prediction[0])
print("Flower name : ", iris.target_names[prediction[0]])
```

```python
import pandas as pd
from sklearn.feature_selection import SelectKBest,f_classif
```

```python
df=pd.read_csv("patient.csv")
```

```python
df["Disease"]=df["Disease"].map({"No":0, "Yes":1})
```

```python
x = df[["Age","BP","Chol","Sugar","BMI"]]
```

```python
y = df["Disease"]
```

```python
selector = SelectKBest(score_func=f_classif,k=3)
x_new = selector.fit_transform(x,y)
```

```python
print("Feature Scores:")
for feature, score in zip(x.columns, selector.scores_):
    print(feature,":",score)
```
