```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv("temperature.csv",encoding="latin1")
print(df)
```

```python
plt.figure(figsize=(16,4))
plt.plot(df["Date"],df["Temperature (°C)"],color="blue",marker="o")
plt.title("Daily Temperature - Line Chart")
plt.xlabel("Date")
plt.ylabel("Temperature (C)")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10,4))
plt.scatter(df["Date"],df["Temperature (°C)"],color="peru")
plt.title("Daily Temperature - Line Chart")
plt.xlabel("Date")
plt.ylabel("Temperature (C)")
plt.xticks(rotation=60)
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10,10))
plt.fill_between(df["Date"],df["Temperature (°C)"],color="pink")
plt.title("Daily Temperature - Line Chart")
plt.xlabel("Date")
plt.ylabel("Temperature (C)")
plt.xticks(rotation=60)
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

sections = ["sci", "Lit", "history", "technology", "commerce"]
books = [450, 380, 250, 520, 310]

plt.figure(figsize=(10, 4))
plt.bar(sections, books, color=["red", "blue", "orange", "yellow", "green"])

plt.title("Sections and Books")
plt.xlabel("section")
plt.ylabel("books")

# Properly indented loop block
for i in range(len(sections)):
    plt.text(i, books[i] + 5, str(books[i]), ha="center")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

np.random.seed(1)
member_id = range(1, 101)
age = np.random.randint(18, 61, 100)

df = pd.DataFrame({"member id": member_id, "age": age})

print(df.head())

plt.figure(figsize=(8, 4))
plt.hist(df["age"], bins=12, color="red", edgecolor="black")

# Corrected titles and axis labels
plt.title("Age Distribution of Members")
plt.xlabel("Age")
plt.ylabel("Frequency (Number of Members)")

plt.show()
```

```python
import matplotlib.pyplot as plt

transport = ["bus", "train", "car", "bike", "bicycle"]
employee = [120, 80, 60, 90, 30]

plt.figure(figsize=(6, 6))
plt.pie(
    employee,
    labels=transport,
    autopct="%1.1f%%",
    colors=["skyblue", "red", "pink", "lightgreen", "gold"],  # Replaced white with gold
    startangle=90,
    wedgeprops={"edgecolor": "black", "linewidth": 1}  # Adds crisp borders
)

plt.title("Preferred Mode of Transportation")
plt.show()
```

```python
import matplotlib.pyplot as plt

cities = ["Pune", "Mum", "C", "D", "E", "F"]
population = [5, 8, 12, 15, 20, 25]
pollution = [40, 48, 55, 63, 72, 85]
green_cover = [350, 300, 250, 200, 180, 150]

# Bar Graph
plt.figure(figsize=(7, 4))
plt.bar(cities, pollution, color="brown")
plt.title("Pollution Level Bar Chart")
plt.xlabel("Cities")
plt.ylabel("Pollution")
plt.show()

# Bubble Plot
plt.figure(figsize=(7, 5))
plt.scatter(
    population,
    pollution,
    s=green_cover,
    color="green",
    alpha=0.6,
    edgecolors="black"
)

# Fixed indentation and centered text alignment
for i in range(len(cities)):
    plt.text(population[i], pollution[i], cities[i], ha="center", va="center")

plt.title("Population vs Pollution")
plt.xlabel("Population (lakhs)")
plt.ylabel("Pollution")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Mocking df for runtime test
df = pd.DataFrame({"Survived": [0, 0, 0, 1, 1]})

Survival = df["Survived"].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(Survival,
        labels=["Not-Survived", "Survived"],
        autopct="%1.1f%%",
        colors=["skyblue", "red"],
        startangle=90)
plt.title("Survival Distribution")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# 1. Load Iris dataset
iris = load_iris()

# 2. Create DataFrame
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["target"] = iris.target  # Added target to color-code by species

# 3. Create Scatter Plot
plt.figure(figsize=(10, 6))
plt.scatter(
    df["sepal length (cm)"],
    df["sepal width (cm)"],
    c=df["target"],        # Colors points based on species class (0, 1, 2)
    cmap="viridis",        # Applies a color map for distinct species colors
    edgecolor="k",
    alpha=0.8
)

# 4. Title and Labels
plt.title("Sepal Length vs Width")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.grid(True)

# 5. Display graph
plt.show()
```
