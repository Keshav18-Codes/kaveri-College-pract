```python
import pandas as pd
import matplotlib.pyplot as plt

sales = [
    2860, 3294, 3130, 3095, 3638, 4169, 2466, 32380, 2330, 3482,
    4135, 4919, 2130, 3685, 2769, 11110, 3515, 4853, 4433, 25000
]

df = pd.DataFrame({"Daily Sales": sales})

plt.figure(figsize=(6, 6))
plt.boxplot(
    df["Daily Sales"],
    patch_artist=True,
    boxprops=dict(facecolor="lightblue", color="navy"),
    flierprops=dict(marker="o", markerfacecolor="red", markersize=8)  # Highlights outliers clearly
)

# Fixed spelling typo in title
plt.title("Distribution of Daily Sales")
plt.ylabel("Sales")
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target
df["Species"] = df["Species"].map({
    0: "Setosa",
    1: "Versicolor",
    2: "Virginica"
})
feature = iris.feature_names
plt.figure(figsize=(12, 8))
for i, feature in enumerate(feature):
    plt.subplot(2, 2, i + 1)
    
    data = [
        df[df["Species"] == "Setosa"][feature],
        df[df["Species"] == "Versicolor"][feature],
        df[df["Species"] == "Virginica"][feature]
    ]
    plt.boxplot(data,
                labels=["Setosa", "Versicolor", "Virginica"],
                patch_artist=True)
    plt.title(feature)
    plt.ylabel("Length / Width(cm)")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "month": ["jan", "feb", "march", "april", "may", "jun", "july", "aug", "sep", "oct", "nov", "dec"],
    "temp": [12, 22, 18, 11, 85, 95, 65, 20, 50, 45, 30, 12],
    "humi": [10, 20, 14, 15, 16, 6, 8, 9, 7, 22, 3, 10],
    "rainf": [12, 8, 99, 62, 1, 5, 10, 10, 11, 21, 14, 15],
    "windsp": [12, 44, 55, 66, 88, 77, 99, 11, 22, 33, 10, 12]
}

df = pd.DataFrame(data)
corr = df.drop("month", axis=1).corr()

plt.figure(figsize=(6, 5))
sns.heatmap(corr,
            annot=True,
            cmap="coolwarm",
            linewidths=0.5)
plt.title("Correlation Heatmap of Weather Parameters")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram

data = {
    "Product": ["p1", "p2", "p3", "p4", "p5", "p6"],
    "Price": [500, 550, 2000, 2100, 800, 850],
    "Rating": [4.0, 4.5, 5.5, 6.0, 6.5, 7.0],
    "Sales": [100, 120, 150, 200, 250, 300]
}

df = pd.DataFrame(data)
x = df[["Price", "Rating", "Sales"]]

z = linkage(x, method="ward")

plt.figure(figsize=(8, 5))
dendrogram(z, labels=df["Product"].values)
plt.title("Dendrogram of products")
plt.xlabel("Products")
plt.ylabel("Distance")
plt.show()
```

```python
# a
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("mpg1.csv")
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.boxplot(df["MPG"])
plt.title("Box plot of MPG")
plt.ylabel("MPG")

plt.subplot(1, 2, 2)
plt.boxplot(df["Horsepower"])
plt.title("Box Plot Of Horsepower")
plt.tight_layout()
plt.show()

# b
corr = df.select_dtypes(include=['number']).corr()
plt.figure(figsize=(7, 6))
sns.heatmap(corr, annot=True, cmap="flare", linewidths=0.5)
plt.title("Correlation Heat Map")
plt.show()

# c
Q1 = df["Horsepower"].quantile(0.25)
Q3 = df["Horsepower"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Fixed the syntax error by adding the proper conditional check (< and >)
outliers = df[(df["Horsepower"] < lower) | (df["Horsepower"] > upper)]
print("Horsepower Outliers")
print(outliers)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib_venn import venn2
```

```python
#a
df = pd.read_csv("movies.csv")

# Extract titles as sets for each platform
netflix = set(df[df["Platform"] == "Netflix"]["Title"])
amazon = set(df[df["Platform"] == "Amazon Prime"]["Title"])

plt.figure(figsize=(8, 8))
venn2([netflix, amazon], set_labels=("Netflix", "Amazon Prime"))
plt.title("Netflix vs Amazon Prime Movies")
plt.show()

#b
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud

df = pd.read_csv("movies.csv")

# Combine all genre text, removing missing values if present
text = " ".join(df["Genre"].dropna())

wc = WordCloud(
    width=700,
    height=400,
    background_color="white"
).generate(text)

plt.figure(figsize=(8, 5))
plt.imshow(wc, interpolation="bilinear")  # 'bilinear' makes word clouds look smoother
plt.axis("off")
plt.title("Genre Word Cloud")
plt.show()

#c
import pandas as pd
from collections import Counter

df = pd.read_csv("movies.csv")

# Combine genres into a single string and count frequencies
text = " ".join(df["Genre"].dropna())
words = text.split()
count = Counter(words)

# Corrected closing parenthesis
print(count)

# Optional: Print the top 10 most common genres nicely
print("\nTop Genres:")
for genre, freq in count.most_common(10):
    print(f"{genre}: {freq}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import squarify

# Load data
df = pd.read_csv("cities.csv")

plt.figure(figsize=(10, 6))
squarify.plot(
    sizes=df["Population"],
    label=df["City"],
    alpha=0.8,
    pad=True  # Optional: Adds spacing between blocks for readability
)

plt.title("City-wise Population Distribution")
plt.axis("off")
plt.show()


#b
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv("cities.csv")

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# Adjusted marker size 's' to a reasonable value
ax.scatter(
    df["Population"],
    df["GDP"],
    df["Literacy_Rate"],
    s=100,             # Reduced size so points remain readable
    c="skyblue",       # Added explicit color
    edgecolors="black" # Added border to distinguish overlapping points
)

ax.set_xlabel("Population")
ax.set_ylabel("GDP")
ax.set_zlabel("Literacy Rate")
plt.title("3D Scatter Plot")

plt.show()
```

```python
import folium

# a
import pandas as pd
import folium

df = pd.read_csv("earthquakes.csv")
m = folium.Map(location=[20, 0], zoom_start=2)

for i in range(len(df)):
    folium.Marker(
        location=[df.loc[i, "latitude"], df.loc[i, "longitude"]]
    ).add_to(m)

m.save("Earthquake_Map.html")
print("Map created successfully!")

# b
df = pd.read_csv("earthquakes.csv")
m = folium.Map(location=[20, 0], zoom_start=2)

for i in range(len(df)):
    # Changed folium.Marker to folium.CircleMarker because radius and fill parameters are for circle markers
    folium.CircleMarker(
        location=[df.loc[i, "latitude"], df.loc[i, "longitude"]],
        radius=df.loc[i, "mag"] * 2,
        color="red",
        fill=True,
        fill_color="orange",
        fill_opacity=0.7,
        popup=f"Magnitude: {df.loc[i, 'mag']}"
    ).add_to(m)

m.save("Earthquake_Magnitude_Map.html")
print("Magnitude map created successfully!")
```
