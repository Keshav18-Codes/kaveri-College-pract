```python
Name=["Tanmay","pooja","diya","keshav"]
age=[20,22,21,29]
plt.bar(Name,age,color=['blue','red','yellow','pink'],edgecolor='black',alpha=0.7)
for i in range(len(age)):
    plt.text(i,age[1],age[i],ha='center')
plt.show()
```

Q2. Draw a stack bar graph of the following data.

```python
boys=[10,5,7,8]
girls=[20,22,21,20]
subjects=["maths","s","b","c"]
plt.bar(subjects,boys,label='boys')
plt.bar(subjects,girls,bottom=boys,label='girls')
plt.legend()
plt.show()
```

```python
Name=["Tanmay","pooja","diya","keshav"]
age=[20,22,21,29]
exp=(0.1,0.2,0.2,0.1)
plt.pie(age,labels=Name,autopct='%1.1f%%',explode=exp,shadow=True)
plt.title("This is a pie diagram")
plt.legend()
plt.axis("equal")
plt.show()
```
