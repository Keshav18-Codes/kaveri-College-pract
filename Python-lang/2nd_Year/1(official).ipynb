Q.3) use Python and plot graph for function 2x-4 and use library scipy and define its functions

```python
#New library import
from scipy.optimize import fsolve
```

```python
#find roots for f(x)
f= lambda x: 2*x-4
root=fsolve(f,x0=0)
print("Root of  the given function is", root)

#Define Roots
x=np.linspace(0,5,10)
y=2*x-4

#Plotting
plt.plot(x,y,label='f(x)=2x-4')
plt.plot(root,f(root),'r*',label=f"root of the function {root}",zorder=50)
plt.axhline(0,color='blue',linestyle='--')
plt.axvline(2,color='blue',linestyle='--')
plt.legend()
plt.grid(color='black')
plt.show()
```

Q.5) use Python and plot graph for function y=sin(x) and use library scipy and define its functions

```python
#Find roots
def f(x):
    return np.sin(x)

root1=fsolve(f,-3.14)[0]
root2=fsolve(f,0)[0]
root3=fsolve(f,3.14)[0]
roots=[root1,root2,root3]

#Define Roots
x=np.linspace(-2*np.pi,2*np.pi,5000)
y=f(x)

#Plotting 
plt.plot(x,y,label='f(x)=sin(x)')
plt.scatter(roots,[f(root) for root in roots], color = 'red', zorder=5, label='Roots (f(x)=0)')
#plt.plot([root1,root2,root3],[0,0,0],'r*',label=f"root of the function {root}")
plt.axhline(0,color='blue',linestyle='--')
plt.axvline(0,color='blue',linestyle='--')
plt.legend()
plt.grid(True,color='black')
plt.show()
```
