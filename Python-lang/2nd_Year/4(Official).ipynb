Create a forward and backward difference table

```python
import numpy as np
import pandas as pd
```

```python
x=[0,1,2,3,4]
y=[0,1,4,9,16]
n=len(x)
diff_table=np.zeros((n,n))
diff_table[:,0]=y
for j in range(1,n):
    for i in range(n-j):
        diff_table[i][j]=diff_table[i+1][j-1]-diff_table[i][j-1]
df=pd.DataFrame(diff_table,columns=[f"d^{i}" if i>0 else 'y' for i in range(n)])
df.insert(0,'x',x)
df.round(2)
```
