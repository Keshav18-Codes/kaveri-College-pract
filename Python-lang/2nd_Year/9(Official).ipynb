create a python program to generate lagrange polynomial

```python
import numpy as np
import sympy as sp
```

```python
def lagrange(x_val,y_val):
    n=len(x_val)
    x=sp.Symbol('x')
    polynomial=0
    for i in range(0,n):
        term=1
        for j in range(0,n):
            if i !=j:
                term=term*(x-x_val[j])/(x_val[i]-x_val[j])
        polynomial= polynomial+y_val[i]*term
    return sp.expand(polynomial)
x_val=[1,3,4,7]
y_val=[1,27,64,125]
lagrange(x_val,y_val)
```
