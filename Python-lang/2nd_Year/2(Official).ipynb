Solving Algebraic & Transcendental Equations using bisection method with Graphical visualization.


```python
import numpy as np
```

Q1)

```python
def f(x):
    return x**3-x**2-2

def bisect(a,b,max_iter=100):
    if f(a)*f(b)>0:
        return none
        print("The value of f(a) & f(b) must be opposite in sign")
    for step in range(max_iter):
        c=(a+b)/2
        fc=f(c)
    if f(a)*fc>0:
        b=c
    else:
        a=c
    return c
a=1
b=2
bisect(a,b)
```

Q2)

```python
from scipy.optimize import bisect
def f(x):
    return x**3-x**2-2
a=1
b=2
root = bisect(f,a,b,xtol=1e-5)
print(root)
```

Q3) 

```python
import matplotlib.pyplot as plt
from scipy.optimize import bisect
def f(x):
    return x**3-x**2-2
a=1
b=2
root = bisect(f,a,b,xtol=1e-5)
print(root)

x=np.linspace(-5,5,100)
y=f(x)
plt.plot(x,y,label='f(x)=x3-x2-2')
plt.plot(root,f(root),'r*',label=f"root of the function {root:4F}",zorder=50)
plt.axhline(0,color='blue',linestyle='--')
plt.axvline(root,color='blue',linestyle='--')
plt.title("Graphical representation of Bisection method")
plt.legend()
plt.grid(color='black')
plt.show()
```
