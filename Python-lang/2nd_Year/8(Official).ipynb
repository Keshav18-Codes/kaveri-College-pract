Lagranges interpolation formula defined using python


```python
import numpy as np
import math
```

```python
def lagrange(x_val,y_val,x):
    n=len(x_val)
    result=0
    for i in range(n):
        term=y_val[i]
        for j in range(n):
            if i !=j:
                term=term*(x-x_val[j])/(x_val[i]-x_val[j])
        result= result+term
    return result
x_val=[1,2,3,4,7]
y_val=[2,4,8,16,128]
x=5
lagrange(x_val,y_val,x)
```
