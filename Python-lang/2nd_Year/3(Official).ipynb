Q.Solving algebraic and transcendental equations using regular falsi method or newton raphson method 

Q1.X^2-2

```python
import numpy as np
```

```python
def f(x):
    return x**2-2
def regula(a,b,max_iter=100,tol=1e-5):
    if f(a)*f(b)>0:
        return None
    for i in range(max_iter):
        c=(a*f(b)-b*f(a))/(f(b)-f(a))
        fc=f(c)
    if fc*f(a)>0:
        a=c
    else:
        b=c
    return c
a=1
b=2
regula(a,b)
```

Q.4 Using newton raphson method

```python
def f(x):
    return x**2-2
    #return x**5+5*x+1
def g(x):
    return 2*x
def newton(x0,max_iter=10,xtol=1e-6):
    for i in range(max_iter):
        fx=f(x0)
        gx=g(x0)
        if gx==0:
            return None
            print("Function does't converges")
        x1=x0-fx/gx
        x0=x1
    return x0
x0=-0.5
root=newton(x0)
print(f"The Approximate root of given function using Newton raphson method is {root}")
```
