```python
import numpy as np
def divided(x_val,y_val,x):
    n=len(x_val)
    diff_table=np.zeros((n,n))
    diff_table[:,0]=y_val
    for j in range(1,n):
        for i in range(n-j):
            diff_table[i][j]=(diff_table[i+1][j-1]-diff_table[i][j-1])/(x_val[i+j]-x_val[i])
    result=y_val[0]
    term=1
    for j in range(1,n):
        term=term*(x-x_val[j-1])
        result=result+term*diff_table[0][j]
    return result
x_val=[2,4,9,10]
y_val=[4,56,711,980]
x=3
divided(x_val,y_val,x)
```
