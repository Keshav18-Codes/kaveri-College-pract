```python
import numpy as np
import matplotlib.pyplot as plt 
from scipy.optimize import fsolve
```

```python
def f(x):
    return x**2-4
```

```python
#Generate x values for plotting
x=np.linspace(-5,5,400)
y=f(x)
```

```python
#Find roots: where f(x)=0
def equation_to_solve(x):
    return f(x)

root1=fsolve(equation_to_solve, -2)[0]
root2=fsolve(equation_to_solve, 2)[0]
roots=[root1, root2]
```

```python
#Plotting
plt.figure(figsize=(8,6))
plt.plot(x,y,label=r'$f(x)=x^2-4$', color='blue',linewidth=2)

#Mark the roots (where f(x) = 0)
plt.scatter(roots,[f(root) for root in roots], color = 'red', zorder=5, label='Roots (f(x)=0)')

#Draw x-axis (y=0) for reference
plt.axhline(0, color='black', linewidth=0.7)
plt.axvline(0, color='black', linewidth=0.7)

#Annotate the roots
plt.annotate(f'x = {root1:.2f}',(root1,0), textcoords="offset points",xytext=(-20,10), ha='center')
plt.annotate(f'x = {root2:.2f}',(root2,0), textcoords="offset points",xytext=(20,10), ha='center')             

#Customize the plot
plt.title('Graph of $f(x) = x^2-4$ and Its Roots', fontsize=14)
plt.xlabel('x',fontsize=12)
plt.ylabel('f(x)',fontsize=12)
plt.grid(True, linestyle='--',alpha=0.6)
plt.legend()
plt.tight_layout()

#Show the plot
plt.show()
```
