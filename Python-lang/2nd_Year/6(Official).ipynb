Use python to create newton gregory formula for forward and backward interpolation

```python
import numpy as np
import math
```

1.

```python
x=[0,1,2,3,4]
y=[0,1,8,27,64]
x_unk=1.5
def forward_interpolation(x,y,x_unk):
    h=x[1]-x[0]
    n=len(x)
    diff=np.zeros((n,n))
    diff[:,0]=y
    for j in range(1,n):
        for i in range(n-j):
            diff[i][j]=diff[i+1][j-1]-diff[i][j-1]
    u=(x_unk-x[0])/h
    u_term=1
    result=y[0]
    for j in range(1,n):
        u_term=u_term*(u-(j-1))
        result=result+(u_term*diff[0][j])/math.factorial(j)
    return result
forward_interpolation(x,y,x_unk)
print(f"This is the final result for given number {forward_interpolation(x,y,x_unk)}")

            
```

```python
x=list(map(int,input('Enter list').split()))
y=list(map(int,input('Enter list').split()))
x_unk=float(input('Enter The number to interpolate'))
def backward_interpolation(x,y,x_unk):
    h=x[1]-x[0]
    n=len(x)
    diff=np.zeros((n,n))
    diff[:,0]=y
    for j in range(1,n):
        for i in range(n-1,j-1,-1):
            diff[i][j]=diff[i][j-1]-diff[i-1][j-1]#for backward
            #diff[i][j]=diff[i+1][j-1]-diff[i][j-1]#for forward
            u=(x_unk-x[-1])/h
    u_term=1
    result=y[-1]
    for j in range(1,n):
        u_term=u_term*(u-(j-1))
        result=result+(u_term*diff[-1][j])/math.factorial(j)
    return result
backward_interpolation(x,y,x_unk)
print(f"This is the final result for given number {backward_interpolation(x,y,x_unk)}")           
```
