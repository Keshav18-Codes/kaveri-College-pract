```python
def Range2(f,x0,y0,h,n):
    x,y=x0,y0
    for i in range(n):
        k1=h*f(x,y)
        k2=h*f(x+h,y+k1)
        y=y+0.5*(k1+k2)
        x=x+h
        print(f'step{i+1}:x={x},y={y}')
    return y
def f(x,y):
    return x+y
Range2(f,0,1,0.1,10)
```

```python
def Range4(f,x0,y0,h,n):
    x,y=x0,y0
    for i in range(n):
        k1=h*f(x,y)
        k2=h*f(x+h/2,y+k1/2)
        k3=h*f(x+h/2,y+k2/2)
        k4=h*f(x+h,y+k3)
        y=y+(k1+2*k2+2*k3+k4)/6
        x=x+h
        print(f'step{i+1}:x={x},y={y}')
    return y
def f(x,y):
    return x+y
Range4(f,0,1,0.1,10)
```
