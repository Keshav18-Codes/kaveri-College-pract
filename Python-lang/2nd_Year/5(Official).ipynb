Using python create newton divided difference table 

```python
import numpy as np
import pandas as pd
```

```python
x=[2,4,9,10]
y=[4,56,711,980]
n=len(x)
diff_table=np.zeros((n,n))
diff_table[:,0]=y
for j in range(1,n):
    for i in range(n-j):
        diff_table[i][j]=(diff_table[i+1][j-1]-diff_table[i][j-1])/(x[i+j]-x[i])
df=pd.DataFrame(diff_table,columns=[f"$\Delta$^{i}" if i>0 else 'y' for i in range(n)])
df.insert(0,'x',x)
df.round(2)
```
