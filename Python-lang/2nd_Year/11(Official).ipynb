```python
def euler(f,x0,y0,h,n):
    x,y=x0,y0
    for i in range(n):
        y=y+h*f(x,y)
        x=x+h
        print(f'step{i+1}:x={x},y={y}')
    return y
def f(x,y):
    return 1/x
euler(f,1,2.5,0.5,4)
```
