```python
def simpson13(f,a,b,n):
    h=(b-a)/n
    result=f(a)+f(b)
    term=0
    for i in range(1,n):
        if i%2==0:
            term=term+2*f(a+i*h)
        else:
            term=term+4*f(a+i*h)
    result = result +term
    return h/3*result
def f(x):
    return 1/x
simpson13(f,1,6,6)
```

```python
def simpson38(f,a,b,n):
    h=(b-a)/n
    result=f(a)+f(b)
    term=0
    for i in range(1,n):
        if i%3==0:
            term=term+2*f(a+i*h)
        else:
            term=term+3*f(a+i*h)
    result = result +term
    return (3*(h/8))*result
def f(x):
    return 1/x
simpson38(f,1,6,6)
```
