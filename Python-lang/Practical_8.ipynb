find combined transformation of 2D object and find transformed figures and plot both

```python
from sympy.geometry import*
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
#combined transformation of a line
```

```python
line = Line(Point(1,1), Point(4,4))

rotated = line.rotate(np.pi/4)

scaled = rotated.scale(2, 2)

transformed = scaled.translate(2, 3)

A, B = line.points
C, D = transformed.points

plt.plot([A.x, B.x], [A.y, B.y],label="Original Line", color="purple")

plt.plot([C.x, D.x], [C.y, D.y],label="Combined Transformed Line", color="green")

plt.axis('equal')
plt.title("Combined Transformation of a Line")
plt.grid()
plt.legend()
plt.show()
```

```python
#combined transformation of a triangle
```

```python
triangle = Triangle(Point(1,1), Point(4,1), Point(2,4))


transformed = triangle.rotate(np.pi/2).translate(2, 3).scale(2, 2)

A, B, C = triangle.vertices
x1 = [A.x, B.x, C.x, A.x]
y1 = [A.y, B.y, C.y, A.y]

P, Q, R = transformed.vertices
x2 = [P.x, Q.x, R.x, P.x]
y2 = [P.y, Q.y, R.y, P.y]

plt.plot(x1, y1, color="purple", label="Original Triangle")

plt.plot(x2, y2, color="green", label="Combined Transformed Triangle")

plt.axis('equal')
plt.title("Combined Transformation of a Triangle")
plt.grid()
plt.legend()
plt.show()
```
