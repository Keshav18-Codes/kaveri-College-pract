Plotting original 2D object and transform after applying any of the 2D transformation such as 
Rotation,Reflection,Scaling,Shearing,Translation.

```python
from sympy.geometry import*
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

Q.1)2D transformations on Line Segment

```python
#rotation of line segment
```

```python
line=Line(Point(1,1),Point(4,4))
rotated=line.rotate(np.pi/4)
A,B=line.points
C,D=rotated.points
plt.plot([A.x,B.x],[A.y,B.y],label="Original Line",color="purple")
plt.plot([C.x,D.x],[C.y,D.y],label="Rotated Line(45)",color="green")

plt.axis('equal')
plt.title("Rotate a Point")
plt.grid()
plt.legend()
plt.show()
```

Q.2)2D transformations on a Triangle

```python
#rotation of triangle
```

```python
T=Triangle(Point(0,0),Point(4,0),Point(2,3))
rotated=T.rotate(np.pi/4)

A,B,C=T.vertices
A1,B1,C1=rotated.vertices

plt.plot([A.x,B.x,C.x,A.x],[A.y,B.y,C.y,A.y],label="Original Triangle",color="purple")

plt.plot([A1.x,B1.x,C1.x,A1.x],[A1.y,B1.y,C1.y,A1.y],label="Rotated Triangle",color="green")
plt.axis("equal")
plt.title("Rotate a triangle")
plt.grid()
plt.legend()
plt.show()
```

```python
#reflection of triangle
```

```python
T=Triangle(Point(0,0),Point(4,0),Point(2,3))
reflected=T.reflect(Line(Point(0,0),Point(1,0)))

A,B,C=T.vertices

x1=[A.x,B.x,C.x,A.x]
y1=[A.y,B.y,C.y,A.y]

A1,B1,C1=reflected.vertices
x2=[A1.x,B1.x,C1.x,A1.x]
y2=[A1.y,B1.y,C1.y,A1.y]

plt.plot(x1,y1,label="Original Triangle",color="purple")

plt.plot(x2,y2,label="Reflected Triangle",color="green")

plt.axis("equal")
plt.title("Reflection of a triangle about X-axis")
plt.grid()
plt.legend()
plt.show()
```

```python
#Scaling of a Triangle
```

```python
T=Triangle(Point(0,0),Point(4,0),Point(2,3))
k=0.5

scaled=T.scale(k)
A,B,C=T.vertices
A1,B1,C1=scaled.vertices

plt.plot([A.x,B.x,C.x,A.x],[A.y,B.y,C.y,A.y],label="Original Triangle",color="purple")

plt.plot([A1.x,B1.x,C1.x,A1.x],[A1.y,B1.y,C1.y,A1.y],label="Scaling of a Triangle",color="green")

plt.axis("equal")
plt.title("Scaling of a triangle")
plt.grid()
plt.legend()
plt.show()
```

```python
#Translation of a triangle
```

```python
T=Triangle(Point(0,0),Point(4,0),Point(2,3))
a=3
b=2
translated=T.translate(a,b)

A,B,C=T.vertices
A1,B1,C1=translated.vertices

plt.plot([A.x,B.x,C.x,A.x],[A.y,B.y,C.y,A.y],label="Original Triangle",color="purple")

plt.plot([A1.x,B1.x,C1.x,A1.x],[A1.y,B1.y,C1.y,A1.y],label="Translated Triangle",color="green")

plt.axis("equal")
plt.title("Translation a triangle")
plt.grid()
plt.legend()
plt.show()
```
