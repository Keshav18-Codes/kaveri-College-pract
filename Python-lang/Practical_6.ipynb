Q.2)Reflect a point and shape

```python
#Point
```

```python
P.reflect(Line(Point(1,0),Point(0,1)))
```

```python
#Polygon
```

```python
P=Polygon(Point(0,0),Point(4,0),Point(5,3),Point(1,3))
P.reflect(Line(Point(1,0),Point(0,3)))
```

Q.3)Scaling of a point and shape

```python
#Point
```

```python
P=Point(2,3)
P.scale(2,1)
```

```python
#Polygon
```

```python
poly=Polygon(Point(1,1),Point(4,1),Point(2,3),Point(2,4))
poly.scale(2,2)
```

Q.4)Shearing of a point and shape

```python
#Point
```

```python
P=Point(3,4)
Sh=Matrix([[1,2,0],[2,1,0],[0,0,1]])
P.transform(Sh)
```

```python
#Polygon
```

```python
shape=Polygon((0,0),(4,0),(2,3))
shx=1
sheared_shape=Polygon(*[(x+shx*y,y) for x,y in shape.vertices])
print(sheared_shape)
```

Q.5)Translation of a Point and Shape

```python
#Point
```

```python
P=Point(2,3)
P_tran=P.translate(5,4)
print("Translated Point:",P_tran)
```

```python
shape=Triangle((0,0),(4,0),(2,3))
translated_shape=shape.translate(5,4)
print("Translated Shape:",translated_shape)
```
