Plotting of 2D Geometrical objects like Points,Line Segments,Triangles and other Polygons.

```python
from sympy.geometry import*
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

Q.1)Draw a Point P

```python
P=Point(2,3)

plt.scatter(P.x,P.y)
plt.text(P.x+0.1,P.y+0.1,"P(2,3)")
plt.axhline(0)
plt.axvline(0)
plt.grid()
plt.show()
```

Q.2)Draw a Line Segment AB

```python
A=Point(0,0)
B=Point(4,4)

plt.plot([A.x,B.x],[A.y,B.y],marker='o')
plt.text(A.x,A.y,"A")
plt.text(B.x,B.y,"B")
plt.grid()
plt.show()
```

Q.3)

```python
line=Line(Point(1,1),Point(3,5))

A,B=line.points

plt.plot([A.x,B.x],[A.y,B.y])
plt.grid()
plt.axis('equal')
plt.show()
```

Draw a Triangle

```python
A=Point(0,0)
B=Point(4,0)
C=Point(2,3)

x=[A.x,B.x,C.x,A.x]
y=[A.y,B.y,C.y,A.y]

plt.plot(x,y,marker='o')
plt.fill(x,y,alpha=0.9,color="orange")
plt.title("Triangle")
plt.grid()
plt.show()
```

Draw a Polygon

```python
P1=Point(0,0)
P2=Point(4,0)
P3=Point(5,3)
P4=Point(1,3)

x=[P1.x,P2.x,P3.x,P4.x,P1.x]
y=[P1.y,P2.y,P3.y,P4.y,P1.y]

plt.plot(x,y,marker='*',color="red")
plt.fill(x,y,alpha=0.3,color="green")
plt.title("Polygon")
plt.grid()
plt.show()
```

Draw a Polygon

```python
square=RegularPolygon(Point(0,0),4,4)

vertices=square.vertices
x=[v.x for v in vertices]+[vertices[0].x]
y=[v.y for v in vertices]+[vertices[0].y]

plt.plot(x,y,marker='*',color="red")
plt.fill(x,y,alpha=0.6,color="purple")
plt.title("Regular Polygon")
plt.grid()
plt.axis('equal')
plt.show()
```

Translate the Line Segment

```python
line=Line(Point(1,1),Point(4,4))
translated=line.translate(2,1)
A,B=line.points
C,D=translated.points
plt.plot([A.x,B.x],[A.y,B.y],label="Original Line")
plt.plot([C.x,D.x],[C.y,D.y],label="Translated Line")

plt.axis('equal')
plt.grid()
plt.legend()
plt.show()
```

Rotate a Point

```python
line=Line(Point(1,1),Point(4,4))
rotated=line.rotate(np.pi/4)
A,B=line.points
C,D=rotated.points
plt.plot([A.x,B.x],[A.y,B.y],label="Original Line",color="purple")
plt.plot([C.x,D.x],[C.y,D.y],label="Rotated Line(45)",color="green")

plt.axis('equal')
plt.title("Rotate a Point")
plt.grid()
plt.legend()
plt.show()
```

Scaling

```python
P=Point(2,3)
S=P.scale(2,1)

plt.scatter(P.x,P.y,label="Original")
plt.scatter(S.x,S.y,label="Scaled")
plt.grid()
plt.legend()
plt.axis('equal')
plt.title("Scaling of Point")
plt.show()
```

Shearing

```python
P=Point(2,3)
Sh=Point(P.x + P.y,P.y)

plt.scatter(P.x,P.y,label="Original",marker='*')
plt.scatter(Sh.x,Sh.y,label="Sheared",marker='*')
plt.grid()
plt.legend()
plt.axis('equal')
plt.title("Shearing of Point")
plt.show()
```

Reflection

```python
P=Point(2,3)
y_axis=Line(Point(0,0),Point(0,1))
R=P.reflect(y_axis)

plt.scatter(P.x,P.y,label="Original",color="green")
plt.scatter(R.x,R.y,label="Reflected",color="brown")
plt.grid()
plt.legend()
plt.axis('equal')
plt.title("Reflection of Point")
plt.show()
```

Rotation

```python
P=Point(2,3)
Rot=P.rotate(np.pi/2)

plt.scatter(P.x,P.y,label="Original",color="red")
plt.scatter(Rot.x,Rot.y,label="Rotated",color="black")
plt.grid()
plt.legend()
plt.axis('equal')
plt.title("Rotation of Point")
plt.show()
```
