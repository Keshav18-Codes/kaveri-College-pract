Finding area,perimeter,midpointof line segment or centroid of 2D objects.point of intersections of two objects and angles between them etc.

Q.1)Area

```python
from sympy.geometry import*
```

```python
A,B,C=Point(0,0),Point(4,0),Point(2,3)
t=Triangle(A,B,C)
t.area
```

```python
p=Polygon(Point(0,0),Point(4,0),Point(5,3),Point(1,3))
p.area
```

Q.2)Perimeter

```python
t.perimeter
```

```python
p.perimeter
```

Q.3)Midpoint

```python
A,B=Point(1,2),Point(5,6)
seg=Segment(A,B)
seg.midpoint
```

Q.4)Centroid of 2D objects

```python
t.centroid
```

```python
p.centroid
```

Q.5)Intersection of two lines.

```python
l1=Line(Point(0,0),Point(4,4))
l2=Line(Point(0,4),Point(4,0))

l1.intersection(l2)
```

Q.6)Intersection of Line and Circle

```python
c=Circle(Point(0,0),5)
l=Line(Point(-6,0),Point(6,0))

l.intersection(c)
```

Q.7)Angle between two lines

```python
l1=Line(Point(0,0),Point(1,0))
l2=Line(Point(0,0),Point(1,1))

l1.angle_between(l2)
```

Q.8)Distance between objects

```python
Point(1,2).distance(Point(4,6))
```

Q.9)Distance between Point and Line

```python
l=Line(Point(0,0),Point(4,0))
p=Point(2,3)

l.distance(p)
```
